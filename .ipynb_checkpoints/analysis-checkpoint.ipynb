# Task 2: Time Series Analysis & Decomposition

```python
!pip install statsmodels
```

```python
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.stattools import adfuller

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
monthly_sales = (
    df.groupby('Order Date')['Sales']
      .sum()
      .resample('ME')
      .sum()
)
```

```python
monthly_sales.head()
```

```python
plt.figure(figsize=(14,6))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    linewidth=2
)

plt.title("Monthly Sales Trend (2014–2017)")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.grid(True)

plt.show()
```

Monthly sales exhibit an overall increasing trend over the four-year period. Several recurring peaks indicate the presence of seasonal buying patterns, especially toward the end of each year.

```python
decomposition = seasonal_decompose(
    monthly_sales,
    model='additive',
    period=12
)
```

```python
fig = decomposition.plot()

fig.set_size_inches(14,10)

plt.show()
```

- Trend :- 
The trend component shows that overall sales increased steadily over the four-year period, indicating business growth.
- Seasonal :- The seasonal component reveals recurring yearly patterns, with certain months consistently producing higher sales.
- Residual :- The residual component captures random fluctuations that are not explained by the trend or seasonality.
- Overall :- The decomposition confirms that sales are influenced by both long-term growth and seasonal demand, making time-series forecasting models suitable.

```python
result = adfuller(monthly_sales)

print("ADF Statistic :", result[0])
print("p-value :", result[1])
print("Critical Values :")

for key, value in result[4].items():
    print(key, ":", value)
```

```python
if result[1] < 0.05:
    print("The time series is stationary.")
else:
    print("The time series is non-stationary.")
```

A stationary time series has statistical properties such as mean and variance that remain constant over time. Most forecasting models, including SARIMA, assume stationarity for accurate predictions. The ADF test is used to determine whether the series is stationary.

```python
monthly_sales_diff = monthly_sales.diff().dropna()
```

```python
plt.figure(figsize=(14,5))

plt.plot(
    monthly_sales_diff.index,
    monthly_sales_diff.values
)

plt.title("Differenced Monthly Sales")
plt.xlabel("Date")
plt.ylabel("Differenced Sales")
plt.grid(True)

plt.show()
```

```python
result_diff = adfuller(monthly_sales_diff)

print("ADF Statistic :", result_diff[0])
print("p-value :", result_diff[1])

for key, value in result_diff[4].items():
    print(key, ":", value)
```

After applying first-order differencing, the time series became stationary. This transformed series is appropriate for statistical forecasting models such as SARIMA.

Interpretation: The ADF test returned a p-value of 0.000278, which is less than the significance level of 0.05. Therefore, we reject the null hypothesis and conclude that the monthly sales time series is stationary. Since the data is already stationary, differencing is not required for further analysis, although it has been demonstrated for learning purposes.

# Task 3 – Sales Forecasting

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    mean_absolute_percentage_error
)

import numpy as np
import matplotlib.pyplot as plt
```

```python
monthly_sales = (
    df.groupby('Order Date')['Sales']
      .sum()
      .resample('ME')
      .sum()
)
```

```python
train = monthly_sales[:-3]
test = monthly_sales[-3:]

print("Training Samples :", len(train))
print("Testing Samples :", len(test))
```

## Model 1 :- SARIMA

```python
(p,d,q) = (1,1,1)

(P,D,Q,m) = (1,1,1,12)
```

```python
sarima_model = SARIMAX(
    train,
    order=(1,1,1),
    seasonal_order=(1,1,1,12),
    enforce_stationarity=False,
    enforce_invertibility=False
)

sarima_result = sarima_model.fit()

print(sarima_result.summary())
```

```python
forecast = sarima_result.get_forecast(steps=3)

forecast_values = forecast.predicted_mean

confidence_intervals = forecast.conf_int()
```

```python
mae = mean_absolute_error(test, forecast_values)

rmse = np.sqrt(
    mean_squared_error(test, forecast_values)
)

mape = mean_absolute_percentage_error(
    test,
    forecast_values
)

print("MAE :", mae)
print("RMSE :", rmse)
print("MAPE :", mape)
```

SARIMA Evaluation: The SARIMA model achieved an MAE of 19,244.49, an RMSE of 19,950.07, and a MAPE of 20.53%. These metrics indicate that the model captures the overall sales trend reasonably well and provides a solid statistical baseline for comparison with Prophet and XGBoost.

```python
plt.figure(figsize=(12,6))

plt.plot(
    train.index,
    train,
    label="Training"
)

plt.plot(
    test.index,
    test,
    label="Actual",
    linewidth=3
)

plt.plot(
    forecast_values.index,
    forecast_values,
    label="Forecast",
    linewidth=3
)

plt.fill_between(
    confidence_intervals.index,
    confidence_intervals.iloc[:,0],
    confidence_intervals.iloc[:,1],
    alpha=0.2
)

plt.title("SARIMA Forecast")

plt.xlabel("Date")

plt.ylabel("Sales")

plt.legend()

plt.grid(True)

plt.show()
```

```python
final_model = SARIMAX(
    monthly_sales,
    order=(1,1,1),
    seasonal_order=(1,1,1,12),
    enforce_stationarity=False,
    enforce_invertibility=False
)

final_result = final_model.fit()
```

```python
future_forecast = final_result.get_forecast(steps=3)

future_values = future_forecast.predicted_mean

future_ci = future_forecast.conf_int()

print(future_values)
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    monthly_sales.index,
    monthly_sales,
    label="Historical Sales"
)

plt.plot(
    future_values.index,
    future_values,
    color="red",
    linewidth=3,
    label="3-Month Forecast"
)

plt.fill_between(
    future_ci.index,
    future_ci.iloc[:,0],
    future_ci.iloc[:,1],
    alpha=0.3
)

plt.title("Future Sales Forecast (SARIMA)")

plt.xlabel("Date")

plt.ylabel("Sales")

plt.legend()

plt.grid(True)

plt.show()
```

## Business Interpretation

The SARIMA model was trained using monthly sales data with yearly seasonality. A train-test split was used to evaluate forecasting performance. The model generated forecasts for the next three months along with confidence intervals, providing an estimate of future sales while accounting for uncertainty. The evaluation metrics (MAE, RMSE, and MAPE) quantify the model's prediction accuracy and will later be compared against Prophet and XGBoost to select the best-performing forecasting approach.

## Model 3 – XGBoost

```python
import pandas as pd
import numpy as np

from xgboost import XGBRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    mean_absolute_percentage_error
)

import matplotlib.pyplot as plt
```

```python
xgb_df = monthly_sales.reset_index()

xgb_df.columns = ['Date', 'Sales']
```

```python
for lag in range(1, 13):
    xgb_df[f'lag_{lag}'] = xgb_df['Sales'].shift(lag)

xgb_df = xgb_df.dropna()

xgb_df.head()
```

```python
train = xgb_df.iloc[:-3]
test = xgb_df.iloc[-3:]
```

```python
X_train = train.drop(columns=['Date', 'Sales'])
y_train = train['Sales']

X_test = test.drop(columns=['Date', 'Sales'])
y_test = test['Sales']
```

```python
model = XGBRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)

model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_test)
```

```python
mae_xgb = mean_absolute_error(y_test, predictions)

rmse_xgb = np.sqrt(
    mean_squared_error(y_test, predictions)
)

mape_xgb = mean_absolute_percentage_error(
    y_test,
    predictions
)

print("MAE :", mae_xgb)
print("RMSE :", rmse_xgb)
print("MAPE :", mape_xgb)
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    test['Date'],
    y_test,
    marker='o',
    linewidth=3,
    label='Actual'
)

plt.plot(
    test['Date'],
    predictions,
    marker='o',
    linewidth=3,
    label='Predicted'
)

plt.title("XGBoost Forecast")

plt.xlabel("Date")

plt.ylabel("Sales")

plt.legend()

plt.grid(True)

plt.show()
```

```python
comparison_df = pd.DataFrame({
    'Model': ['SARIMA', 'Prophet', 'XGBoost'],
    'MAE': [mae, mae_prophet, mae_xgb],
    'RMSE': [rmse, rmse_prophet, rmse_xgb],
    'MAPE': [mape, mape_prophet, mape_xgb]
})

comparison_df = comparison_df.sort_values(by='MAPE')

comparison_df
```

```python
comparison_df.to_csv("model_comparison.csv", index=False)
```

Three forecasting approaches—SARIMA, Prophet, and XGBoost—were evaluated using MAE, RMSE, and MAPE. SARIMA consistently achieved the lowest error values, demonstrating the best forecasting performance on the Superstore monthly sales data. Prophet produced comparable results but with slightly higher errors, while XGBoost performed the weakest due to the limited size of the time-series dataset. Based on these results, SARIMA was selected as the final forecasting model.

## Task 4: Category-wise & Region-wise Forecasting

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def sarima_forecast(data, title):
    """
    Trains a SARIMA model and forecasts the next 3 months.
    """

    # Create monthly sales series
    monthly = (
        data.groupby('Order Date')['Sales']
        .sum()
        .resample('ME')
        .sum()
    )

    # Train model
    model = SARIMAX(
        monthly,
        order=(1,1,1),
        seasonal_order=(1,1,1,12),
        enforce_stationarity=False,
        enforce_invertibility=False
    )

    result = model.fit(disp=False)

    # Forecast
    forecast = result.get_forecast(steps=3)

    future = forecast.predicted_mean

    confidence = forecast.conf_int()

    # Plot
    plt.figure(figsize=(12,5))

    plt.plot(monthly.index,
             monthly.values,
             label="Historical Sales",
             linewidth=2)

    plt.plot(future.index,
             future.values,
             color="red",
             marker='o',
             linewidth=3,
             label="Forecast")

    plt.fill_between(
        confidence.index,
        confidence.iloc[:,0],
        confidence.iloc[:,1],
        alpha=0.2
    )

    plt.title(title)

    plt.xlabel("Date")

    plt.ylabel("Sales")

    plt.legend()

    plt.grid(True)

    plt.show()

    print("\nForecast for Next 3 Months")

    display(future.to_frame(name="Forecast Sales"))

    return future
```

```python
technology = df[df['Category']=="Technology"]

technology_forecast = sarima_forecast(
    technology,
    "Technology Sales Forecast"
)
```

```python
furniture = df[df['Category']=="Furniture"]

furniture_forecast = sarima_forecast(
    furniture,
    "Furniture Sales Forecast"
)
```

```python
office = df[df['Category']=="Office Supplies"]

office_forecast = sarima_forecast(
    office,
    "Office Supplies Sales Forecast"
)
```

```python
category_forecast = pd.DataFrame({
    "Technology": technology_forecast.values,
    "Furniture": furniture_forecast.values,
    "Office Supplies": office_forecast.values
})

category_forecast.index = technology_forecast.index

category_forecast
```

## Task 5 – Anomaly Detection

```python
monthly_sales = (
    df.groupby('Order Date')['Sales']
      .sum()
      .resample('ME')
      .sum()
)
```

```python
anomaly_df = monthly_sales.to_frame(name='Sales')

anomaly_df.head()
```

```python
anomaly_df['Rolling Mean'] = anomaly_df['Sales'].rolling(window=3).mean()

anomaly_df['Rolling Std'] = anomaly_df['Sales'].rolling(window=3).std()

anomaly_df.head()
```

```python
anomaly_df['Z-Score'] = (
    anomaly_df['Sales'] - anomaly_df['Sales'].mean()
) / anomaly_df['Sales'].std()
```

```python
anomaly_df['Anomaly'] = anomaly_df['Z-Score'].abs() > 2

anomaly_df.head()
```

```python
anomalies = anomaly_df[anomaly_df['Anomaly']]

print("Number of anomalies detected:", len(anomalies))

anomalies
```

```python
plt.figure(figsize=(14,6))

plt.plot(
    anomaly_df.index,
    anomaly_df['Sales'],
    label='Monthly Sales',
    linewidth=2
)

plt.scatter(
    anomalies.index,
    anomalies['Sales'],
    color='red',
    s=100,
    label='Anomalies',
    zorder=5
)

plt.title("Monthly Sales with Detected Anomalies")

plt.xlabel("Date")

plt.ylabel("Sales")

plt.grid(True, linestyle='--', alpha=0.5)

plt.legend()

plt.show()
```

```python
plt.figure(figsize=(14,6))

plt.plot(
    anomaly_df.index,
    anomaly_df['Sales'],
    label='Sales',
    linewidth=2
)

plt.plot(
    anomaly_df.index,
    anomaly_df['Rolling Mean'],
    label='Rolling Mean',
    linewidth=2
)

plt.fill_between(
    anomaly_df.index,
    anomaly_df['Rolling Mean'] - anomaly_df['Rolling Std'],
    anomaly_df['Rolling Mean'] + anomaly_df['Rolling Std'],
    alpha=0.2,
    label='Rolling Std'
)

plt.title("Rolling Mean and Standard Deviation")

plt.xlabel("Date")

plt.ylabel("Sales")

plt.grid(True)

plt.legend()

plt.show()
```

```python
anomaly_df.to_csv("anomaly_detection_results.csv")
```

## Bussiness insights 

- Monthly sales were analyzed using the Z-score method with a threshold of ±2.
- The identified anomalies represent months where sales deviated significantly from the overall sales pattern.
- These unusual observations may correspond to promotional campaigns, seasonal demand, special events, or unexpected market conditions.
- Detecting anomalies helps the business investigate exceptional performance and improve future planning, inventory management, and marketing strategies.

```python
from sklearn.ensemble import IsolationForest
```

```python
weekly_sales = (
    df.groupby("Order Date")["Sales"]
      .sum()
      .resample("W")
      .sum()
)

weekly_df = weekly_sales.to_frame(name="Sales")

weekly_df.head()
```

```python
iso = IsolationForest(
    contamination=0.05,
    random_state=42
)

weekly_df["Isolation"] = iso.fit_predict(
    weekly_df[["Sales"]]
)
```

```python
weekly_df["Isolation"] = weekly_df["Isolation"].map({
    1:"Normal",
    -1:"Anomaly"
})

weekly_df.head()
```

```python
plt.figure(figsize=(14,6))

plt.plot(
    weekly_df.index,
    weekly_df["Sales"],
    label="Weekly Sales"
)

anomaly = weekly_df[
    weekly_df["Isolation"]=="Anomaly"
]

plt.scatter(
    anomaly.index,
    anomaly["Sales"],
    color="red",
    s=80,
    label="Anomaly"
)

plt.title("Isolation Forest Anomaly Detection")

plt.legend()

plt.grid(True)

plt.show()
```

```python
weekly_df["Rolling Mean"] = (
    weekly_df["Sales"]
    .rolling(4)
    .mean()
)

weekly_df["Rolling Std"] = (
    weekly_df["Sales"]
    .rolling(4)
    .std()
)
```

```python
weekly_df["Z"] = (
    weekly_df["Sales"]-
    weekly_df["Rolling Mean"]
)/weekly_df["Rolling Std"]

weekly_df["Z Anomaly"] = weekly_df["Z"].abs()>2

weekly_df.head()
```

```python
plt.figure(figsize=(14,6))

plt.plot(
    weekly_df.index,
    weekly_df["Sales"],
    label="Weekly Sales"
)

plt.scatter(
    weekly_df[
        weekly_df["Z Anomaly"]
    ].index,

    weekly_df[
        weekly_df["Z Anomaly"]
    ]["Sales"],

    color="orange",
    s=80,
    label="Z-score Anomaly"
)

plt.title("Z-score Anomaly Detection")

plt.grid(True)

plt.legend()

plt.show()
```

```python
print(
    "Isolation Forest anomalies:",
    (weekly_df["Isolation"]=="Anomaly").sum()
)

print(
    "Z-score anomalies:",
    weekly_df["Z Anomaly"].sum()
)
```

```python
weekly_df.to_csv(
    "weekly_anomaly_detection.csv"
)
```

Two anomaly detection techniques were applied to identify unusual sales patterns. The Z-score method detected statistical outliers based on deviations from the mean, while Isolation Forest identified anomalies using an unsupervised machine learning approach. Comparing both methods provides a more comprehensive understanding of abnormal sales behavior, enabling the business to investigate exceptional demand periods and improve future forecasting and inventory planning.

## Task 6 – Product Demand Segmentation using Clustering

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
monthly_subcat = (
    df.groupby([
        'Sub-Category',
        pd.Grouper(key='Order Date', freq='ME')
    ])['Sales']
    .sum()
    .reset_index()
)

monthly_subcat.head()
```

```python
feature_df = (
    monthly_subcat
    .groupby('Sub-Category')
    .agg(
        Total_Sales=('Sales', 'sum'),
        Avg_Order_Value=('Sales', 'mean'),
        Sales_Volatility=('Sales', 'std')
    )
)
```

```python
growth = (
    monthly_subcat
    .sort_values(['Sub-Category', 'Order Date'])
)

growth['Growth'] = (
    growth.groupby('Sub-Category')['Sales']
    .pct_change()
)

growth_rate = (
    growth.groupby('Sub-Category')['Growth']
    .mean()
)

feature_df['Growth_Rate'] = growth_rate

feature_df = feature_df.fillna(0)

feature_df.head()
```

```python
scaler = StandardScaler()

scaled_features = scaler.fit_transform(feature_df)
```

```python
wcss = []

for k in range(1, 8):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(scaled_features)

    wcss.append(model.inertia_)
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    range(1,8),
    wcss,
    marker='o'
)

plt.title("Elbow Method")

plt.xlabel("Number of Clusters")

plt.ylabel("WCSS")

plt.grid(True)

plt.show()
```

```python
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

kmeans.fit(scaled_features)
```

```python
feature_df["Cluster"] = kmeans.labels_

feature_df.head()
```

```python
feature_df.to_csv("demand_segmentation.csv")
```

```python
pca = PCA(n_components=2)

pca_features = pca.fit_transform(scaled_features)

pca_df = pd.DataFrame(
    pca_features,
    columns=["PC1","PC2"]
)

pca_df["Cluster"] = feature_df["Cluster"].values
```

```python
plt.figure(figsize=(8,6))

for cluster in sorted(pca_df["Cluster"].unique()):

    cluster_data = pca_df[pca_df["Cluster"] == cluster]

    plt.scatter(
        cluster_data["PC1"],
        cluster_data["PC2"],
        label=f"Cluster {cluster}",
        s=100
    )

plt.title("Demand Segmentation using PCA")

plt.xlabel("Principal Component 1")

plt.ylabel("Principal Component 2")

plt.legend()

plt.grid(True)

plt.show()
```

```python
cluster_summary = (
    feature_df
    .groupby("Cluster")
    .mean(numeric_only=True)
)

cluster_summary
```

```python
cluster_names = {
    0: "High Demand",
    1: "Medium Demand",
    2: "Low Demand"
}

feature_df["Demand Segment"] = (
    feature_df["Cluster"]
    .map(cluster_names)
)

feature_df
```

```python
feature_df.to_csv(
    "demand_segmentation.csv",
    index=True
)
```

Demand Segmentation Analysis

Product sub-categories were segmented using K-Means clustering based on total sales, average order value, sales volatility, and growth rate. The Elbow Method was used to determine the optimal number of clusters, and PCA was applied for visualization. These segments help identify groups of products with similar demand characteristics, enabling better inventory planning, targeted marketing, and resource allocation.
